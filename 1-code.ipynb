```python
# Databricks Notebook: Sales Data Pipeline

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum, avg

# Step 1: Create Spark Session
spark = SparkSession.builder.appName("SalesPipeline").getOrCreate()

# Step 2: Create Sample Data (Simulating Raw Data)
data = [
    (1, "Laptop", 2, 50000),
    (2, "Mobile", 5, 20000),
    (3, "Tablet", 3, 15000),
    (4, "Laptop", 1, 52000),
    (5, "Mobile", 2, 21000)
]

columns = ["order_id", "product", "quantity", "price","Ramu"]

df = spark.createDataFrame(data, columns)

print("Raw Data:")
display(df)

# Step 3: Transformation (Business Logic)
# Calculate total revenue per product

df_transformed = df.withColumn("total_price", col("quantity") * col("price"))

result = df_transformed.groupBy("product").agg(
    sum("total_price").alias("total_revenue"),
    avg("price").alias("avg_price")
)

print("Transformed Data:")
display(result)

```
